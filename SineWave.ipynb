```python
# built-in python
range_iterable = range(2, 5, 1) # range makes an iterable (https://www.w3schools.com/python/python_iterators.asp)

# range makes a list of all integers from arg 1 up to and excluding arg 2,
# you can change its increments with arg 3
# https://docs.python.org/2/library/functions.html

# we can convert it to a  list using something called "coercian" which means forced conversion of one data type to another
range_list = list(range_iterable)
range_list
```

```python
# if this function was not already built in, we could make it on our own
#          arg 1,  arg 2,  kwarg
def _range(first, second, iter=1):
  assert first < second
  i = first
  result = []
  while i < second:
    result.append(i)  # [2, 3, 4, 5]
    i += iter  # this is the same as: i = i + iter

  return result
```

So now our tests show our function works how we'd expect within a tolerable margin of error. But how can we make it safer? We know that x, A, and b should all be numbers.
We also know the value of our return should be a floating point number (decimal)

Python does not check types for us, but it can sometimes mutate (change) the types to force them to work with certain operations.

```python
# for example
input = 3
output = input/2
print(
    f'type of {input} is {type(input)} '
    f'but type of {output} is {type(output)}) '
    '\nOur int has mutated to a float because the division coerces ints into floats before it runs'
    )
```

How can we check the types in our own function to make sure we do not have unintended consequences?

```python
list_to_check = [True, 1, 1.0, '1']
```

```python
type(list_to_check) # type of a list is a list, what if we want the type of something in the list?
```

```python
# use a for-loop
for item in list_to_check:
  # f-string = make a string from your variables and function calls
  print(f'type of {item} is {type(item)}')
  # can we coerce (force) it to be a floating point (decimel) type?
  print(f'coerced type of {item} is {type(float(item))}')
  print('------------\n') # \n = new line
```

```python
# can also use a list-comprhension (aka a simple, one line for-loop)
[type(item) for item in list_to_check]
```

If we want to **protect** our function from **variables** that should **already be floats** we can check if they are floats already

```python
# check if all items are floats this is the most safe, but may be too strict
# for example, you may want to let them be integers as well since they are
# commonly used as arguments for your sine function

if not all([type(item) == float for item in list_to_check]): # all() checks if all items in a list are True
  raise TypeError( # raise error forces the program to stop and give you a message
      f"The following items are not floats:\n"
      f"{[item for item in list_to_check if type(item) != float]}"
  )

# read more about errors here: https://docs.python.org/3/tutorial/errors.html
```

We also might try to see if we can **coerce** them to floats (aka, force python to try to make them floats).

This is least likely to raise an error, but much more likely to be dangerous (as in, it can allow unintended consequences)

```python
# let's add a new value to our list to check using append:
list_to_check.append('one')  # adds 'one' to the end of list_to_check, note that this changes the list in place rather than returns a new variable
list_to_check
```

```python
# check if all items can be turned into floats

# make an empty list to store the results of our check
check_results = []
for item in list_to_check:

  try: # try-except statements execute try, and then execute the except if the try raises an error
    float(item)  # coercian to float, will raise an error if not possible
    check_results.append( (item, True) ) # appends a tuple (value1, value2) to our list
  except:
    check_results.append( (item, False) )

print(check_results)
```

We can now see that not only ints, but bools and strings that contain only numeric characters will happily be coerced into floats if we force them too.

How might coercian cause uninteded side effects? Here is an example, what if intead of using a list of tuples to store our checks like in the last code
block, we used a dictionary? First, a reminder: a dictionary is like a list, but there is no order, and we use keys to access values instead of indexes and slices

```python
# for example, if we wanted to store a dictionary of animals and check whether
# they were mammals, we could do:
is_mammal = {
    'cat': True,
    'dog': True,
    'bird': False,
    'fish': False
}

# and we can update a dict we already made like this:
is_mammal['bunny'] = True

# get get the values like this
print(is_mammal['bird'])
```

Now, let's rewrite our type-checker to use a dict and see what happens!

```python

# check if all items can be turned into floats using a dictionary

# make an empty dictionary
check_results = {} # read more about dictionaries here: https://www.geeksforgeeks.org/python/python-dictionary/
for item in list_to_check:

  try: # try-except statements execute try, and then execute the except if the try raises an error
    float(item)  # coercian to float, will raise an error if not possible
    check_results[item] = True
  except:
    check_results[item] = False

print(check_results)
```

Do you see any problems? What happened to 1.0 and 1?

Here is another example of coercian doing bad things:

```python
arr = np.array([1, '2', 3])
print(arr)
print(arr.dtype)
```

See what happened? Numpy cannot have mixed- datatypes. It converted every item in the list to the type with the largest memory requirement. in this case a string

Here's a funny example:

```python
def is_successful(x):
    return x > 10

score = 8
bonus = is_successful(score) * 5  # Oops
print(bonus)
```

Python will happily treat bools as 1s and 0s, because technically that is what they are! In practice, you probably would never want it to, and if your multiplying a bool with a number, you'd probably want your function to tell you something has gone horribly wrong.

# Let's apply what we've learned to make a safer sine_wave function

We will also add a few more parameters to make it behave like a sound wave

```python
# let's apply what we learned to make a safe sine_wave functoin:
# y(x) = A·sin[k·(x - Δx) + φ] for the array of x values provided.
def sine_wave(
    x: float, # position
    A: float = 1, # amplitude
    k: float = 1, # wave number
    dx: float = .1, # change in x
    phi: float = 0 # phase
) -> float:
  """
  This function computes the 1-dimensional sine wave:
    y(x) = A·sin[k·(x - dx) + phi]

  Args:
    x: position of the observer
  Kwargs:
    A: amplitide
    k: wave number
    dx: is the displacement of x (i.e. change in x relative to the starting point)
    phi: phase shift

  Returns:
    y: magnitude of the sine wave at position x
  """
  ok_types = [float, int]
  bad_types = [bool, str]
  for item in [x, A, k, dx, phi]:
    print(f'{item}: {type(item)}')
    if type(item) in bad_types:
      raise TypeError(
          f'value {item} is an invalid type ({type(item)})\n'
          f'Okay types are {ok_types}'

          )
    item = float(item)
    # f-string f before string means i assign variables in the string
    if type(item) not in ok_types:
      raise TypeError(f'value {item} is NOT coercible to a float')

  y = A * np.sin(k*(x-dx) + phi)
  print(type(y))
  # this checks if y is any kind of float
  if not np.issubdtype(y, np.floating):
    raise TypeError('Something went wrong! y is not a number')
```

```python
# y(x) = A·sin[k·(x - Δx) + φ] for the array of x values provided.
def sine_wave(
    x: float, # position
    A: float = 1, # amplitude
    k: float = 1, # wave number
    dx: float = .1, # change in x
    phi: float = 0 # phase
) -> float:
  """
  This function computes the 1-dimensional sine wave:
    y(x) = A·sin[k·(x - dx) + phi]

  Args:
    x: position of the observer
  Kwargs:
    A: amplitide
    k: wave number
    dx: is the displacement of x (i.e. change in x relative to the starting point)
    phi: phase shift

  Returns:
    y: magnitude of the sine wave at position x
  """
  ok_types = [float, int] # types we want to allow
  bad_types = [bool, str] # types we want to explicitly disallow
  for item in [x, A, k, dx, phi]:
    if type(item) in bad_types:
      raise TypeError(
          f'value {item} is an invalid type ({type(item)})\n'
          f'Okay types are {ok_types}')

    failed = False
    try:
      item = float(item)  # coerce to a float BEFORE using numpy
      if type(item) not in ok_types:
        failed = True
    except:
      failed = True
    if failed:
      raise TypeError(f'value {item} was NOT coercible to a number')

  y = A * np.sin(k*(x-dx) + phi)

  # this checks if y is any kind of float
  if not np.issubdtype(y, np.floating):
    raise TypeError('Something went wrong! y is not a float')

  return y
```
